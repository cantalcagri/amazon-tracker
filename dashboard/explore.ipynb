## Units Sold Per Seller (after 2nd day of data)

```python
sales = pd.read_sql("""
    SELECT u.calc_date, p.asin, s.seller_name, s.fulfillment,
           u.inv_yesterday, u.inv_today, u.units_sold, u.oos_sold
    FROM fact_daily_units_sold u
    JOIN dim_product p ON u.product_id = p.product_id
    JOIN dim_seller s ON u.seller_id = s.seller_id
    WHERE u.is_new_seller = 0 AND u.units_sold > 0
    ORDER BY u.calc_date DESC, u.units_sold DESC
""", conn)
sales
```

## Daily Summary Per ASIN

```python
agg = pd.read_sql("""
    SELECT a.agg_date, p.asin, a.bsr_rank,
           ROUND(a.avg_price_weighted,2) AS avg_price,
           a.total_sellers, a.fba_sellers, a.fbm_sellers,
           a.total_units_sold
    FROM agg_product_daily a
    JOIN dim_product p ON a.product_id = p.product_id
    ORDER BY a.agg_date DESC
""", conn)
agg
```
